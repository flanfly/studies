```python
import polars as pl

df = (
    pl.read_csv('sector-rotation.trail2.csv', has_header=True)
    .pivot(on='name',values='data')
    .with_columns(
        pl.col("parameters").str.json_decode(dtype=pl.Struct([
            pl.Field("signal", pl.Utf8),
            pl.Field("period", pl.Utf8),
            pl.Field("stop_long", pl.Utf8),
            pl.Field("stop_short", pl.Utf8),
            pl.Field("max_long", pl.Utf8),
            pl.Field("max_short", pl.Utf8),
            pl.Field('daily_exit', pl.Utf8),
            pl.Field("leverage", pl.Utf8),
            pl.Field("hard_stop_long", pl.Utf8),
            pl.Field("hard_stop_short", pl.Utf8),
        ]))
    )
    .unnest('parameters')
    .sort(['sortino','cagr'], descending=True)
)


df
```
