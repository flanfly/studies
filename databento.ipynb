```python
import databento as db

from dotenv import load_dotenv
import os

load_dotenv()

client = db.Historical(os.getenv("DATABENTO_APIKEY"))
data = client.timeseries.get_range(
    dataset="OPRA.PILLAR",
    symbols=["XLK.OPT"],
    stype_in='parent',
    schema='ohlcv-1d',
    start="2022-05-02T00:00:00",
    end="2022-05-04T00:00:00",
)
```

```python
data.to_df()
```
